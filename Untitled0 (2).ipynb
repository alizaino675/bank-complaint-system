```python
import re
import re


def cleaning_text(text: str):

    text = text.lower()
    text = re.sub(r'x{2,}/x{2,}/x{4,}', '', text)
    text = re.sub(r'xx/xx/xxxx', '', text)

    text = re.sub(r'\bx{2,}\b', '', text)
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    text = re.sub(r'http\S+|www\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)


    text = re.sub(r'\s+', ' ', text).strip()

    return text


df['Consumer_complaint_narrative_preprocessed'] = df['Consumer complaint narrative'].apply(cleaning_text)

```

```python
import pandas as pd
from transformers import AutoTokenizer, AutoModelForSequenceClassification, Trainer, TrainingArguments




tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")



model = AutoModelForSequenceClassification.from_pretrained("bert-base-uncased", num_labels=df['label'].unique().shape[0])
```

```python
from datasets import Dataset
from transformers import DataCollatorWithPadding

dataset = Dataset.from_pandas(df)

def tokenize(example):
  return tokenizer(example['text'], truncation=True)

tokenized_dataset = dataset.map(tokenize, batched=True)

tokenized_dataset = tokenized_dataset.remove_columns(['text'])

data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

tokenized_dataset = tokenized_dataset.train_test_split(test_size=0.2, seed = 42)
```

```python
from transformers import EarlyStoppingCallback
training_args = TrainingArguments(
    output_dir="./results",
    eval_strategy="steps",
    learning_rate=1e-5,
    per_device_train_batch_size=16,
    num_train_epochs=3,
    weight_decay=0.01,
    eval_steps=50,              # Evaluate every single global step
    logging_steps=50,

    load_best_model_at_end=True,
    metric_for_best_model="eval_loss",
    greater_is_better=False,

)

# 6. بدء التدريب
```

```python
from transformers import Trainer
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = logits.argmax(axis=-1)
    return {"accuracy": (predictions == labels).mean()}

trainer = Trainer(
    model=model,
    args=training_args,
    data_collator=data_collator,
    train_dataset=tokenized_dataset["train"],
    eval_dataset=tokenized_dataset["test"],
    compute_metrics=compute_metrics,
    callbacks=[EarlyStoppingCallback(early_stopping_patience=3)]
)

trainer.train()
```

```python
text = "My mortgage payment was processed twice this month, and I am still waiting for the bank to refund the extra payment."

import torch


inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True)







inputs = {k: v.to(model.device) for k, v in inputs.items()}

with torch.no_grad():
    outputs = model(**inputs)

prediction = outputs.logits.argmax(dim=1).item()

print(prediction)
```
